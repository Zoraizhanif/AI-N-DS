# Question 7

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, x, /): ")

if operator == "+":
    print("Result:", num1 + num2)
elif operator == "-":
    print("Result:", num1 - num2)
elif operator == "x":
    print("Result:", num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Cannot divide by zero")
else:
    print("Invalid operator")
```

```python
salary = float(input("Enter salary: "))

if salary > 600000:
    tax = salary * 0.05
    gross_salary = salary - tax
    print("Tax:", tax)
    print("Gross Salary:", gross_salary)
else:
    print("Gross Salary:", salary)
```

# Question 13

```python
salary = float(input("Enter salary: "))

if salary > 2500000:
    tax = salary * 0.15
elif salary >= 1500000 and salary <= 2500000:
    tax = salary * 0.10
elif salary >= 500000 and salary < 1500000:
    tax = salary * 0.05
else:
    tax = 0

gross_salary = salary - tax

print("Tax:", tax)
print("Gross Salary:", gross_salary)
```
